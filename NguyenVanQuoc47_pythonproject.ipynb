```python
import pandas as pd
# Tạo dữ liệu cho 10 sinh viên
data = {
    "Họ và tên": [
        " Mai Phạt Sáu Nghìn",
        "Lò Vi Sóng",
        "Phan Bá Đạo",
        "Đồng Hồ Cát",
        "Đường Thẳng Tắp",
        "Nguyễn Thị Kim Cương",
        "Võ Lâm Truyền Kỳ",
        "Trần Như Nhộng",
        "Sống Văn Hoá",
        "Cao Bằng Lạng Sơn",
    ],
    "Điểm chuyên cần": [9, 8, 10, 7, 8, 9, 6, 10, 7, 8],
    "Điểm giữa kỳ": [8, 7, 9, 6, 8, 9, 7, 8, 6, 9],
    "Điểm cuối kỳ": [9, 8, 8, 7, 9, 8, 6, 9, 7, 10]
}
#1 Tạo DataFrame
df = pd.DataFrame(data)
#2  Tính điểm tổng kết
df["Tong_ket"] = (
    0.2 * df["Điểm chuyên cần"]
    + 0.3 * df["Điểm giữa kỳ"]
    + 0.5 * df["Điểm cuối kỳ"]
)
# Đổi chỉ mục từ 0-9 thành 1-10
df.index = range(1, 11)

# Yêu cầu 3: Xếp loại sinh viên
def xep_loai(diem):
    if diem >= 8.5:
        return "Giỏi"
    elif diem >= 7.0:
        return "Khá"
    elif diem >= 5.0:
        return "Trung bình"
    else:
        return "Yếu"

df["Xep_loai"] = df["Tong_ket"].apply(xep_loai)
# Yêu cầu 4: Thống kê

# 1. Tính điểm tổng kết trung bình của lớp
diem_trung_binh = df["Tong_ket"].mean()
print("\n1. Điểm tổng kết trung bình của lớp:",
      round(diem_trung_binh, 2))

# 2. Tìm sinh viên có điểm tổng kết cao nhất
diem_cao_nhat = df["Tong_ket"].max()
sinh_vien_cao_nhat = df[df["Tong_ket"] == diem_cao_nhat]

print("\n2. Sinh viên có điểm tổng kết cao nhất:")
print(sinh_vien_cao_nhat[["Họ và tên", "Tong_ket"]])

# 3. Tìm sinh viên có điểm tổng kết thấp nhất
diem_thap_nhat = df["Tong_ket"].min()
sinh_vien_thap_nhat = df[df["Tong_ket"] == diem_thap_nhat]

print("\n3. Sinh viên có điểm tổng kết thấp nhất:")
print(sinh_vien_thap_nhat[["Họ và tên", "Tong_ket"]])

# 4. Đếm số sinh viên đạt (điểm tổng kết >= 5)
so_sinh_vien_dat = (df["Tong_ket"] >= 5).sum()
print("\n4. Số sinh viên đạt:", so_sinh_vien_dat)

# Hiển thị bảng dữ liệu sau khi tính
print(df)

```
